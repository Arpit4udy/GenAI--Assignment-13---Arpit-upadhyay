## DATA CLEANING


```python
import pandas as pd
```

```python
df=pd.read_csv('online food delivery dataset.csv')
df
```

```python
numerical_columns=['Family size','latitude','longitude','Pin code']
categorical_columns=['Gender','Marital Status','Occupation','Monthly Income','Educational Qualifications','Customer Type','Output','Feedback','Unnamed: 13',]
```

```python
df['Family size'].fillna(df['Family size'].mean(),inplace=True)
```

## Removing Duplicate Values

```python
df.duplicated().sum()
```

```python
df=df.drop_duplicates()
```

```python
df.shape
```

## Renaming Columns

```python
df.columns=(df.columns).str.lower().str.replace(' ','_')
```

```python
df.columns
```

## Changing Data Types

```python
df.dtypes
```

```python
df['family_size']=df['family_size'].astype('int32')
df['latitude']=df['latitude'].astype('float32')
df['longitude']=df['longitude'].astype('float32')
df['pin_code']=df['pin_code'].astype('int32')

```

```python
df.to_csv('online food delivery.csv')
```
