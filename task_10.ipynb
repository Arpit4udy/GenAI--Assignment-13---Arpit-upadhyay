```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('online food delivery.csv')
```

## Patterns

```python
sns.countplot(data=df,x='age')
plt.show()
```

```python
df['age'].describe()
```

The Age of customers lies between 18 year to 33 year with average age of 24.6 years.


```python
sns.countplot(data=df,x='feedback')
plt.show()
```

Positive feedback is more common than negative feedback.

## Outliers

```python
sns.boxplot(data=df,x='age')
plt.show()
```

A few unusually high age values are observed and may require further investigation.


```python
sns.boxplot(data=df,x='family_size')
plt.show()
```

No outliers were obseerved in family Size. Most of them lie between 2 and 4.

## Relationships

```python
sns.boxplot(data=df, x='feedback', y='age')

plt.title("Age Distribution by Feedback")
plt.xlabel("Feedback")
plt.ylabel("Age")
plt.show()
```

A negative relationship is observed between age and feedback, where older individuals tend to give more negative feedback compared to younger individuals.

```python
sns.scatterplot(data=df,x='family_size',y='age')
plt.title("Age vs Family_size")
plt.ylabel('Age')
plt.xlabel('Family_size')
plt.show()
```

No strong relationship is observed between age and family size.
