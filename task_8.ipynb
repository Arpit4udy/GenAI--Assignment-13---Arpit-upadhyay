## Univariate Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('online food delivery.csv')
```

```python
sns.histplot(data=df,x='age')
plt.show()
```

```python
sns.kdeplot(data=df,x='age')
plt.show()
```

```python
sns.countplot(data=df,x='family_size')
plt.show()
```

```python
sns.kdeplot(data=df,x='family_size')
plt.show()
```

```python
sns.countplot(data=df,x='gender')
plt.show()
```

```python
sns.countplot(data=df,x='marital_status')
plt.show()
```

```python
sns.countplot(data=df,x='customer_type')
plt.show()
```

```python
sns.countplot(data=df,x='occupation')
plt.show()
```

```python
sns.countplot(data=df,x='educational_qualifications')
plt.show()
```

```python
sns.countplot(data=df,x='feedback')
plt.show()
```

## other Chart

```python
sns.violinplot(data=df, x="age")
plt.show()
```

```python
counts = df["occupation"].value_counts()

plt.figure(figsize=(6, 6))

plt.pie(
    counts,
    labels=counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Occupations")
plt.show()
```

```python
counts = df["feedback"].value_counts()

plt.figure(figsize=(6, 6))

plt.pie(
    counts,
    labels=counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("FeedBack")
plt.show()
```
