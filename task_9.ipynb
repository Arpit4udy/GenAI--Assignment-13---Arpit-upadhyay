## Bivariate Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('online food delivery.csv')
```

```python
df
```

```python
sns.scatterplot(data=df,x='family_size',y='age')
plt.title("Age vs Family_size")
plt.ylabel('Age')
plt.xlabel('Family_size')
plt.show()
```

```python
corr = df[['age', 'family_size']].corr()

sns.heatmap(corr, annot=True)

plt.show()
```

## Categorical vs Numerical

```python
sns.barplot(data=df, x='gender', y='family_size')

plt.title("Average Family Size by Gender")
plt.show()
```

```python
sns.boxplot(data=df, x='gender', y='occupation')

plt.title("Occupation by Gender")
plt.show()
```

```python
sns.barplot(
    data=df,
    x='educational_qualifications',
    y='family_size'
)

plt.xticks(rotation=45)
plt.title("Average Family Size by Education Qualification")
plt.show()
```

```python
sns.boxplot(data=df, x='feedback', y='age')

plt.title("Age Distribution by Feedback")
plt.xlabel("Feedback")
plt.ylabel("Age")
plt.show()
```
